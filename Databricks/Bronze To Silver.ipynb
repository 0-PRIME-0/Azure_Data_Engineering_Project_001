```python
from pyspark.sql.functions import to_date, col
def convertIntoDate(Spath,Tpath):
    df=spark.read.format("csv")\
                .option("header","true")\
                .option("inferSchema","true")\
                .load(Spath)
    columns=df.columns
    for c in columns:
        if "Date" in c or "date" in c:
            df=df.withColumn(c ,to_date(col(c),"yyyy-MM-dd'T'HH:mm:ss Z"))
    df.write.format("delta").mode("overwrite").save(Tpath)
```

```python
for i in table_name:
    Spath="/Volumes/prdatabricks001_7405611854915130/my_catalog/bronze/SalesLT/"+i
    Tpath="/Volumes/prdatabricks001_7405611854915130/my_catalog/silver/SalesLT/"+i
    convertIntoDate(Spath,Tpath)
```
