#### Apply Columns Name Transform For All Tables

```python
table_name=[]
for i in dbutils.fs.ls("/Volumes/prdatabricks001_7405611854915130/my_catalog/silver/SalesLT/"):
    table_name.append(i.name.split("/")[0])
table_name
```

```python
def transform_col_name(sPath,tPath):
    df=spark.read.format("delta").load(sPath)
    columns = df.columns

    for c in columns:
        nc = c[0]

        for i in range(1, len(c)):
            if c[i].isupper() and c[i-1].islower():
                nc += "_"

            nc += c[i]

        print(nc)
        df = df.withColumnRenamed(c, nc)
    df.write.format("delta").mode("overwrite").save(tPath)
```

```python

for t in table_name:
    sPath="/Volumes/prdatabricks001_7405611854915130/my_catalog/silver/SalesLT/"+t
    tPath="/Volumes/prdatabricks001_7405611854915130/my_catalog/gold/SalesLT/"+t
    transform_col_name(sPath,tPath)
```
